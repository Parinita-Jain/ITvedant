```python
#pattern=re.compile("[\w\s]+")
pattern=re.compile("[a-zA-Z0-9_ ]+")
s="100 cat'S is sitting on a SINGLE mat" # a-za-z0-9_
re.findall(pattern,s)
```

```python
text="Hello WOrld Hello !"
pattern="Hello"
match=re.match(pattern,text)
print(match)
print(match.group())
fa=re.findall(pattern,text) # list and giving all the occurences
print(fa)
```

```python
text="the quick brown fox jumped over a lazy dog FOX fox Fox"
pattern="fox"
fa=re.findall(pattern,text) # list and giving all the occurences, case sensitive
print(fa)
```

```python
text="Hello WOrld Hello !"
pattern="Hello"
fa=re.search(pattern,text) # giving 1st occurences, case sensitive
print(fa.group())
print(fa)
```

```python
text="Contact at 8989898989ghfhgjf9090909090" #-- 6-9{1}0-9{9}
pattern=re.compile("[6-9]{1}[0-9]{9}")
print(re.findall(pattern,text))
```

```python
n=input("Enter phone number : ")
if len(n)==10:
    pattern=re.compile(r"[6-9]{1}[\d]{9}")
    x=re.findall(pattern,n)
    if len(x)!=0:
     print("Valid phone number : ",x)
    else:
        print("Invalid")
else:
    print("Invalid")
    
```

```python
text="the cat sat on a mat mat mat mat mat"
s="hat"
replaced_text=re.sub("mat",s,text)
print(replaced_text)
```

```python
text="the cat sat on a 'mat' ; mat ; mat  : mat ; mat"
print(re.split(";",text))
```

```python
pattern=r"[\w\d.]+@[\w]+\.[\w]{2,3}"
text="my email id is abc@gmail.com rug@gmail.com"
re.findall(pattern,text)
```

```python
pattern=r"[\w\d.]+@[gmail|yahoo]+\.[com|in]+"
text="my email id is abc@gmail.com"
re.findall(pattern,text)
```
