```python
r = requests.get(url)#,auth=('user', 'pass'))
r.status_code
```

```python
soup = BeautifulSoup(r.text, 'html.parser')
soup
```

```python
quote_data=[]
quote_divs=soup.find_all("div",class_="quote")
quote_divs
```

```python
#for i in quote_divs:
quotes_data=[]
for i in quote_divs:
    text=i.find("span",class_="text").text.strip()
    author=i.find("small",class_="author").text.strip()
    x=i.find_all("a",class_="tag")
    tag=[]
    for j in x:
        tag.append(j.text)
    tags=", ".join(tag)
    quotes_data.append({"Quote":text,"Author":author,"Tags":tags})
quotes_data
```

```python
df=pd.DataFrame(quotes_data)
df.head(2)    
```

```python
df.to_csv("Quotes.csv")
```
