```python
sns.boxplot(data=df,y="make",x="price")
```

```python
def identifying_outliers(g):
    q1=g["price"].quantile(0.25)
    q3=g["price"].quantile(0.75)
    iqr=q3-q1
    lw=q1-1.5*iqr
    uw=q3+1.5*iqr
    outliers=g[(g["price"]<lw)|(g["price"]>uw)]
    return outliers["price"]
```

```python
df.groupby("make").apply(identifying_outliers)
```

```python
# replacing outliers by winsorizing method
def replacing_outliers(g):
    q1=g["price"].quantile(0.25)
    q3=g["price"].quantile(0.75)
    iqr=q3-q1
    lw=q1-1.5*iqr
    uw=q3+1.5*iqr
    g["price"]=g["price"].apply(lambda x:min(max(x,lw),uw))
    return g
```

```python
df=df.groupby("make").apply(replacing_outliers).reset_index(drop=True)
```
