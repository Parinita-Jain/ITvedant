```python
# dictionaries
d={
    "student_id":[1,2,3,4,5,6,7,8,9],
    "Name":["April","Autumn","Rose","Lily","Mary","Tom","Jerry","Bob","Harry"],
    "Marks":[89,80,90,88,87,90,99,89,87]
}
df=pd.DataFrame(d)
print(df.head(1))
print(df.tail(4))
print(df.sample(4))
```

```python
d={"id":[1,2,3,4,5],
   "Name":["April","Autumn","Rose","Lily","Mary"],
    "Marks":[89,80,90,88,87]
  }
df1=pd.DataFrame(d)
print(df1)

d={
    "student_id":[1,2,3],
    "age":[13,14,15]
}
df2=pd.DataFrame(d)
print(df2)
df3=pd.merge(df1,df2,left_on="id",right_on="student_id",how="inner")
df3
```
