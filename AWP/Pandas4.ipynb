```python
print(df[['Series or Movie','Boxoffice']].sample(4))
print(df[['Series or Movie','Boxoffice']].dtypes)
df['Boxoffice']=df['Boxoffice'].str.replace("[$,]+","",regex=True).astype(float)
```

```python
len(df[((df['Series or Movie']=="Movie")&(df['Boxoffice']>2000000))])
```

```python
# critically acclaimed movies having imdb score>=8
df['IMDb Score'].dtypes
```

```python
len(df[((df['Series or Movie']=="Movie")&(df['IMDb Score']>=8))])
```

```python
#critically acclaimed movies having imdb score>=8 boxoffice>2000000
len(df[((df['Series or Movie']=="Movie")&(df['IMDb Score']>=8)
       &(df['Boxoffice']>2000000))])
```

```python
# movies that are nominated of highest number of awards >=5.
df.columns
df['Awards Nominated For'].dtypes
min_5_awards=df[df['Awards Nominated For']>=5].sort_values(by="Awards Nominated For",ascending=False)
min_5_awards.head(3)
len(min_5_awards[min_5_awards["Series or Movie"]=="Movie"])
```

```python
# United States series with imdb score>=8
len(df[(
    (df['Series or Movie']=="Series")&
    (df['IMDb Score']>=8) &
    (df['Country Availability']=="United States")
       )])
```
