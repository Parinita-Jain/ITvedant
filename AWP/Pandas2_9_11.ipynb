```python
# grading - 75-100 - return A, 60-74 - B,35-59 - C, F
def grade(x):
    x=x/2
    if x>=75 and x<=100:
        return "A"
    elif x>=60 and x<=74:
        return "B"
    elif x>=35 and x<=59:
        return "C"
    else:
        return "F"
```

```python
#df["Total"].apply(lambda x:grade(x/2))
df["Grade"]=df["Total"].apply(grade)
```
